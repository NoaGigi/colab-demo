# Hello from Colab!

This notebook defines a simple addition function and prints a greeting and the result of adding 2 and 3. Run the code cell in Google Colab; no extra packages are needed.

```python
def add(a, b):
    return a + b

print("Hello from Colab!")
print(add(2, 3))
```

## Check the GitHub connection

Run all cells in order. This check downloads this public notebook from GitHub and tests the addition function. It needs internet access but no extra packages or credentials. It checks public read access; saving changes to GitHub still requires signing in through Colab.

```python
import json
from urllib.request import Request, urlopen

url = 'https://raw.githubusercontent.com/NoaGigi/colab-demo/main/hello_colab.ipynb'
request = Request(url, headers={'User-Agent': 'colab-demo-connection-check'})
with urlopen(request, timeout=20) as response:
    github_notebook = json.load(response)

assert github_notebook['nbformat'] == 4, 'The downloaded file is not a notebook.'
assert add(2, 3) == 5
assert add(-2, 3) == 1
assert add(0, 0) == 0
print('GitHub connection works: downloaded hello_colab.ipynb from NoaGigi/colab-demo.')
print('All addition checks passed!')
```
